### Step 2: 演習用データの生成とファイル操作
RNA-seq の解析を想定した模擬の FASTQ ファイルを作成し、コピー・移動を行います。

```python
%%bash
# [実行1] テスト用FASTQデータの自動生成（4行×3リード = 12行）
cat << 'EOF' > sample_reads.fastq
@READ_001_Actin
ATGCGATCGATCGATCGATCGATC
+
IIIIIIIIIIIIIIIIIIIIIIII
@READ_002_GAPDH
GCTAGCTAGCTAGCTAGCTA
+
IIIIIIIIIIIIIIIIIIIIIIII
@READ_003_Actin
ATCGATCGATCGATCGATCGATCG
+
IIIIIIIIIIIIIIIIIIIIIIII
EOF
```

```python
# ファイルができたか確認(sample_reads.fastq)
!ls -la
```

```python
# [実行2] ファイルの複製 (cp)
!cp sample_reads.fastq backup_reads.fastq
```

```python
# 複製できたか確認（backup_reads.fastqがあるか？）
!ls -la
```

```python
# [実行3] ディレクトリを作成して移動
# ディレクトリ作成(mkdir)
!mkdir -p raw_data
```

```python
# ファイルを移動(mv)
!mv backup_reads.fastq raw_data/
```

```python
# [実行4] 一覧表示で確認
!ls raw_data/
```

```python
# ファイ名変更(mv)
!mv raw_data/backup_reads.fastq raw_data/renamed_backup_reads.fastq
```

```python
#名前が変わったか確認（renamed_backup_reads.fastq）
!ls raw_data/
```
